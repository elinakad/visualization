```python
import numpy as np
import matplotlib.pyplot as plt

Year = np.arange(1982, 2026)

Production = np.array([
    2642, 4491, 4391, 4129, 3600, 4277, 3912, 3742, 4763, 4247,
    3076, 4784, 2815, 3790, 3458, 3318, 4776, 3740, 3338, 3219,
    4824, 3903, 3456, 4055, 3759, 2708, 2081, 2600, 3999, 4093,
    4281, 2886, 4798, 4070, 4028, 2761, 3707, 3612, 3872, 2904,
    3443, 3630, 2673, 3561
])
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 4), facecolor='lightgrey')
plt.suptitle('Honey Industry Trends (1982–2025)', fontsize=16, fontweight='bold', color='black')

# -----------------------
# Total Honey Production
plt.plot(Year, Production, color='black')
plt.title('Total Honey Production', fontweight='bold', color='black')
plt.xlabel('Year', fontweight='bold', color='black')
plt.ylabel('Tonnes', fontweight='bold', color='black')
plt.xticks(rotation=45, color='black')
plt.grid(True, linestyle='--', alpha=0.5, color='black')
ax.set_facecolor('lemonchiffon')

# Caption at the bottom
plt.figtext(0.02, -0.05,
            "Figure 1, Total Honey Production in Canada from 1982 to 2025. Source: Statistics Canada",
            fontsize=8, ha='left', va='bottom', color='black')
```
