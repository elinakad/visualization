```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(618)
x = np.arange(50)
y = np.random.randint (0, 100, 50)
```

```python
fig, ax = plt.subplots (figsize = (5, 3)) 
ax.scatter (x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
ax.hist(y, edgecolor='black')
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total Growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
font1 = {
    'family': 'sans-serif',
    'color': 'blue',
    'fontsize': 20
}

font2 = {
    'family': 'monospace',
    'color': 'green',
    'fontsize': 14
}
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict=font1)
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,
           y,
        marker='*',
        color = "indigo")   
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,
           y,
        marker='*',
        color = 'indigo',
        linestyles = '--',
        linewidth = 2)
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
# color being the new addition
ax.plot(x,y,marker='*', color ='#7425b9', linestyle = '--', linewidth = 2)
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*',
markersize = 12, # NEW ADDITIONS
color = '#7425b9',
linestyle = '--',
linewidth = 2,
markeredgecolor = '#fa9359',
markerfacecolor = '#000000')
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*',
markersize = 12, # NEW ADDITIONS
color = '#7425b9',
linestyle = '--',
linewidth = 2, 
markeredgecolor = '#fa9359',
markerfacecolor = '#000000')
ax.grid(axis = 'y')
ax.grid(axis = 'y', color = 'blue', linewidth = 2, linestyle = '--')
```
